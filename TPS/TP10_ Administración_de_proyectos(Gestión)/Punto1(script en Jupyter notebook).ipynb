```python
def calcular_vp_anualidad(cuota, tasa, periodos):
    return cuota * ((1 - (1 + tasa) ** -periodos) / tasa)

def calcular_vp_futuro(monto_futuro, tasa, periodos):
    return monto_futuro / ((1 + tasa) ** periodos)

tasa = 0.01
costo_mensual = 1000.0

# a. VPN (12 meses)
vpn_a = calcular_vp_futuro(18000, tasa, 12) - calcular_vp_anualidad(costo_mensual, tasa, 12)
print(f"a. Valor Presente Neto (12 meses): ${vpn_a:.2f}")

# b. VPN (15 meses)
vpn_b = calcular_vp_futuro(18000, tasa, 15) - calcular_vp_anualidad(costo_mensual, tasa, 15)
print(f"b. Valor Presente Neto (15 meses): ${vpn_b:.2f}")

# c. Rentabilidad
print(f"c. Rentabilidad (12 meses): {vpn_a / calcular_vp_anualidad(costo_mensual, tasa, 12):.2%}")
print(f"c. Rentabilidad (15 meses): {vpn_b / calcular_vp_anualidad(costo_mensual, tasa, 15):.2%}")

# d. VPN retraso de 6 meses (total 18 meses)
vpn_d = calcular_vp_futuro(18000, tasa, 18) - calcular_vp_anualidad(costo_mensual, tasa, 18)
print(f"d. VPN (18 meses): ${vpn_d:.2f}")

# e. Gestión profesional (Costo + 5%)
vpn_e = calcular_vp_futuro(18000, tasa, 12) - calcular_vp_anualidad(costo_mensual * 1.05, tasa, 12)
print(f"e. VPN con Gestión Profesional: ${vpn_e:.2f}")

# f. Desafío
cuota_maxima = (calcular_vp_futuro(18000, tasa, 15) - vpn_a) / calcular_vp_anualidad(1, tasa, 15)
print(f"f. Costo mensual máximo aceptable: ${cuota_maxima:.2f}")
```
